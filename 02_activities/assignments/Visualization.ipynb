```python
import pandas as pd
import json
import plotly.express as px
```

```python

# Load data
df = pd.read_csv("./data.csv")

# Processing data
df['geometry'] = df['geometry'].apply(json.loads)
df['lon'] = df['geometry'].apply(lambda g: g['coordinates'][0][0])
df['lat'] = df['geometry'].apply(lambda g: g['coordinates'][0][1])
status_map = {0: 'Closed', 1: 'Temporarily Closed', 2: 'Open'}
df['status_label'] = df['Status'].map(status_map)

# Ploting data
fig = px.scatter_mapbox(
    df,
    lat="lat",
    lon="lon",
    color="status_label",
    hover_name="AssetName",
    hover_data=["Reason", "Comments","hours","type", "location_details"],
    zoom=10,
    mapbox_style="open-street-map",
    title="Public Park Washroom Facilities in Toronto"
)

fig.show()
fig.write_html("toronto_park_washrooms_map.html")
```
